## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

I chose ranking as the machine learning task because the goal is to prioritize webpages based on their risk of declining search performance. Instead of simply predicting whether a page will decline or not, the model should rank pages so the SEO team can review the highest-risk pages first. This supports better decision making by helping allocate time and resources to the pages that are most likely to benefit from optimization.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is whether a webpage is showing a declining search trend. I will use the **trend_direction** column as the target, where pages with the value **"down"** represent declining performance. This label comes from an **observed outcome** in the dataset rather than a manually defined rule, making it a suitable proxy for identifying pages that may need SEO attention.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will use **Precision@50** as the main success metric because the goal is to recommend the most important pages for the SEO team to review first. A higher Precision@50 means that most of the top 50 recommended pages are actually showing declining search performance, making the recommendations more useful and reducing time spent reviewing healthy pages.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*


The unit of analysis is a **single webpage**. Each row in the dataset represents one webpage along with its SEO and performance metrics, such as impressions, CTR, average position, content age, and trend direction. These features will be used to help rank pages that may need attention.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Shape of the dataset:", df.shape)

# Display the first 5 rows
df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule cannot reliably identify declining pages because search performance depends on many factors working together. Features such as CTR, impressions, content age, freshness, average position, engagement rate, and trend direction all influence whether a page needs attention.

Machine learning can learn patterns from all of these features at the same time instead of relying on one threshold or simple if-statement. This provides better decision-support for prioritizing which pages should be reviewed by the SEO team.

```python
features = [
    "ctr",
    "impressions_90d",
    "content_age_days",
    "days_since_last_update",
    "avg_position",
    "engagement_rate",
    "trend_direction"
]

print(df[features].head())
```
