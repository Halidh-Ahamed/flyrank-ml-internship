# ML-02 — Research Question and Provisional Lane

This notebook defines my research question and provisional lane for the FlyRank Machine Learning Internship.

## 1. My lane (or freestyle) and why

I chose the Google Search Ranking & Discoverability lane because it focuses on identifying pages that are likely to decline in search performance. By using measurable signals such as impressions, click-through rate (CTR), page age, and content freshness, I hope to help prioritize which pages should be reviewed first. This can support SEO decisions using observable data rather than assumptions.

## 2. The question: decision, action, cost of a wrong call

The goal of this project is to identify web pages that may be at risk of declining search performance so they can be reviewed early. The decision supported by this analysis is which pages should be prioritized for optimization. The SEO or content team can use these recommendations to focus their efforts efficiently. A wrong recommendation could cause valuable time to be spent on healthy pages or allow truly declining pages to be ignored, potentially resulting in reduced search visibility and traffic.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

declining = (df["trend_direction"] == "down").sum()

print("Total pages:", len(df))
print("Pages with declining trend:", declining)
print("Average CTR:", round(df["ctr"].mean(), 3))
```

This project provides decision-support based on observed patterns in the available data. It can help identify pages that may deserve attention for content updates or SEO optimization.

However, this project cannot prove why a page's performance changes, cannot predict Google's algorithm, and does not establish cause-and-effect relationships. The findings should be treated as directional insights rather than guaranteed outcomes.
